```r
### EDIT THIS CODE CELL ###
#TA told me to do this
data_summary_stats <- tidied_data |>
    select(reg_wins, tourney_wins, seed, total_losses) |> #here, I am selecting for my chosen variables
    summarize(across(reg_wins: total_losses, list(max = max, min = min, mean = mean))) #here, I am finding the max, min, and mean for al my chosen variables. note that na.rm = TRUE was not included, as the remove NA step was already completed in the tidying step.
data_summary_stats
```

```r
### EDIT THIS CODE CELL FOR PLOT 1 ###

reg_wins_plot <- tidied_data |>
    ggplot(aes(x= reg_wins)) + #reg_wins was chosen as the numeric variable, and it was plotted on the x-axis.
        geom_histogram(binwidth = 1) + #here, I chose a histogram as the best representation for this visualization, as it is the best choice to show the distribution of 1 quantitative variable. binwidth was tested with many values, but 1 was the most reasonable choice.
        labs(x='Number of Regional Wins', y='Number of Teams with X Regional Wins', title='Distribution of Regional Wins') + #here, I labelled the graph, and the axes.
        theme(text=element_text(size=20)) #here, I changed the size of the text so the plot could be read more easily.
reg_wins_plot
```

```r
### EDIT THIS CODE CELL FOR PLOT 2 ###

category_proportions <- tidied_data |>
    group_by(bid) |>
    summarize(total_numbers = n()) |>
    mutate(proportions = total_numbers/(sum(total_numbers)))
# category_proportions


category_proportion_plot <- category_proportions |>
    ggplot(aes(x=bid, y=proportions)) + #bid was chosen as the x-variable, as it is the only categorical variable. the y-axis is the proportion of each bid type
        geom_bar(stat= 'identity') + #I chose to use a bar plot for this visualization, to best represent the proportion of each bid type
        labs(x='Bid status', y='Proportion of Bid Type (/1)', title='Proportion of Bid Type') + #here, I labelled the graph, and the axes.
        ylim(0, 1) +
        theme(text=element_text(size=20)) #here, I changed the size of the text so the plot could be read more easily.
category_proportion_plot
```

Since we are attempting to predict the bid status (auto or at-large) of a team, it makes the most sense to see the distribution of the total losses, and compare them to at-large versus auto. This way, we can see whether there is a significant difference between the at-large vs auto distributions of this variable. The dual-grid histogram (facet) allows us to compare the two more easily. Here, this plot allows us to see that both at-large and auto bids are clustered around the same values, with at-large having a greater peak.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 1###

reg_losses <- tidied_data |>
    ggplot(aes(x=total_losses, fill = bid)) +
        geom_histogram(binwidth = 1) + #chose a histogram to represent this relationship, best choice because of ability to see the distribution among a categorical variable
        facet_grid(rows=vars(bid)) + #created two grids to compare between each bid type
        labs(x='Number of Total Losses', y='Number of Teams with X Total Losses', title='Distribution of Total Losses per Bid Type', fill = 'Bid Status') + #here, I labelled each section of the graph
        theme(text = element_text(size=20)) #here, I increased the text size so I can read it better (I have bad eyes so this works very well)
reg_losses
```

Once again, we are using a histogram and facet grid design to visualize this relationship. This will be useful when deciding how to represent our final classification. Comparing this with VISUALIZATION 1 allows us to see that Tourney Wins has a far smaller value range compared to Total Losses. From this, we can conclude that we will need to process our data to ensure fairness when performing classification. Otherwise, the reasoning behind the use of this plot is similar to VISUALIZATION 1's.

```r
### EDIT THIS CODE CELL FOR VISUALIZATION 2###

tourn_wins_bid <- tidied_data |>
    ggplot(aes(x=tourney_wins, fill = bid)) +
        geom_histogram(binwidth = 1) + #chose a histogram again, same reason as previous plot
        facet_grid(rows=vars(bid)) + #same reasoning as previous plot
        labs(x='Number of Tourney Wins', y='Number of Teams with X Tourney Wins', title='Distribution of Tourney Wins per Bid Type', fill = 'Bid Status') +
        theme(text = element_text(size=20))
tourn_wins_bid
```
